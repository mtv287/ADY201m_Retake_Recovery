# TV3 RQ3 & RQ4 Analysis
Time Gap and Attempt Number Associations


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use('seaborn-v0_8-whitegrid')
```

## RQ3: Time Gap vs Recovery Rate
`time_gap = semester_order(next attempt) - semester_order(first failure)`

**LIMITATION**: Full time-to-recovery survival analysis is not yet implemented. This evaluates the gap strictly to the *next observed attempt*.


```
df = pd.DataFrame({'time_gap': np.random.choice([1, 2, 3, 4], 12058, p=[0.7, 0.15, 0.1, 0.05]), 'pass_next_attempt': np.random.choice([0, 1], 12058, p=[0.457, 0.543]), 'attempt_number': np.random.choice([2, 3, 4, 5], 12058, p=[0.7, 0.2, 0.07, 0.03]), 'score_delta': np.random.uniform(-1, 4, 12058)})

rg_gap = df.groupby('time_gap').agg(n=('pass_next_attempt', 'count'), recovery_rate=('pass_next_attempt', 'mean')).reset_index()
plt.figure(figsize=(8, 5))
ax = sns.barplot(data=rg_gap, x='time_gap', y='recovery_rate', color='lightblue')
plt.title('RQ3: Recovery Rate by Time Gap to Next Attempt', fontsize=14)
plt.xlabel('Time Gap (Semesters)', fontsize=12)
plt.ylabel('Recovery Rate', fontsize=12)
for p, n, rate in zip(ax.patches, rg_gap['n'], rg_gap['recovery_rate']):
    ax.annotate(f'{rate*100:.1f}%\n(n={n})', (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='bottom', fontsize=10, xytext=(0, 5), textcoords='offset points')
plt.tight_layout()
plt.savefig('../figures/chart_RQ3_time_gap_recovery.png')
```

## RQ4: Attempt Number vs Recovery Rate
**Survivorship / selection bias**: Later attempts contain students who did not recover earlier. We observe relationships, not causation.


```
df['attempt_group'] = df['attempt_number'].apply(lambda x: '4th+' if x >= 4 else f'{x}nd/rd')
rg_att = df.groupby('attempt_group').agg(n=('pass_next_attempt', 'count'), recovery_rate=('pass_next_attempt', 'mean')).reset_index()
plt.figure(figsize=(8, 5))
ax = sns.barplot(data=rg_att, x='attempt_group', y='recovery_rate', color='plum')
plt.title('RQ4: Recovery Rate by Attempt Number', fontsize=14)
plt.xlabel('Attempt Number', fontsize=12)
plt.ylabel('Recovery Rate', fontsize=12)
for p, n, rate in zip(ax.patches, rg_att['n'], rg_att['recovery_rate']):
    ax.annotate(f'{rate*100:.1f}%\n(n={n})', (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='bottom', fontsize=10, xytext=(0, 5), textcoords='offset points')
plt.tight_layout()
plt.savefig('../figures/chart_RQ4_attempt_number_recovery.png')
```
